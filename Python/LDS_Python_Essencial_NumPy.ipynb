## 1. De R para Python: o que muda

| Conceito | R | Python |
|---|---|---|
| Atribuição | `x <- 5` | `x = 5` |
| Primeiro elemento | `x[1]` | `x[0]` |
| Intervalo | `1:5` → 1,2,3,4,5 | `range(1, 6)` → 1,2,3,4,5 (fim **exclusivo**) |
| Vetor | `c(1, 2, 3)` | `[1, 2, 3]` (lista) ou `np.array([1, 2, 3])` |
| Lista nomeada | `list(a = 1, b = 2)` | `{"a": 1, "b": 2}` (dicionário) |
| Blocos | `{ }` | **indentação** (4 espaços) |
| Lógicos (valor único) | `TRUE`, `FALSE`, `&&`, `\|\|`, `!` | `True`, `False`, `and`, `or`, `not` |
| Lógicos (vetorizado) | `&`, `\|`, `!` | `&`, `\|`, `~` (em arrays NumPy e Pandas) |
| Ausente | `NA`, `NULL` | `None` (e `np.nan` em dados numéricos) |
| Pacotes | `library(dplyr)` | `import pandas as pd` |
| Métodos | `toupper(s)` | `s.upper()` (o método "pertence" ao objeto) |
| Cópia | ao modificar, R copia | atribuir **não copia**: duas variáveis podem apontar para o mesmo objeto |

> **As três pegadinhas que mais geram erro:** indexação começando em 0, intervalos com fim exclusivo, e atribuição por referência. Vamos ver cada uma.

````python
# Pegadinha 1: indexação começa em 0
frutas = ["maçã", "banana", "uva", "manga"]
print(frutas[0])    # primeiro
print(frutas[-1])   # último (índice negativo conta do fim)
````

````python
# Pegadinha 2: fim exclusivo
print(list(range(1, 5)))   # 1, 2, 3, 4  (o 5 não entra)
print(frutas[1:3])         # posições 1 e 2
````

````python
# Pegadinha 3: atribuição por referência
a = [1, 2, 3]
b = a          # b NÃO é cópia: aponta para a mesma lista
b.append(99)
print("a:", a)  # a também mudou!

c = a.copy()   # cópia de verdade
c.append(100)
print("a:", a, "| c:", c)
````

> Em R, `b <- a; b[4] <- 99` não altera `a`. Em Python, listas, dicionários, arrays e DataFrames são **mutáveis** — use `.copy()` quando quiser um objeto independente. Isso vai voltar no Pandas.

## 2. Tipos e estruturas de dados

### 2.1 Tipos básicos

````python
idade = 34            # int
altura = 1.78         # float
nome = "Ana"          # str
aprovado = True       # bool
nada = None           # ausência de valor

for v in [idade, altura, nome, aprovado, nada]:
    print(repr(v), "→", type(v).__name__)
````

````python
# Operadores: atenção à divisão
print(7 / 2)    # 3.5  divisão real
print(7 // 2)   # 3    divisão inteira  (R: %/%)
print(7 % 2)    # 1    resto           (R: %%)
print(2 ** 3)   # 8    potência        (R: ^)
````

### 2.2 Strings e f-strings

````python
cidade = "curitiba"
print(cidade.upper(), cidade.title(), len(cidade))
print(cidade.replace("c", "C", 1))
print("ti" in cidade)

# f-string: o equivalente moderno de paste()/sprintf()
media = 7.456
print(f"A média da turma em {cidade.title()} foi {media:.2f}")
````

### 2.3 Listas — sequência ordenada e mutável
Próximas de um `vector`/`list` do R, mas aceitam tipos misturados e **não** são vetorizadas.

````python
notas = [7.5, 8.0, 6.5, 9.0]
notas.append(5.5)          # adiciona no fim
notas.sort()               # ordena no lugar
print(notas, len(notas), sum(notas) / len(notas))
````

````python
# Cuidado: lista NÃO é vetor
print([1, 2, 3] * 2)       # repete a lista!  (em R: 2 4 6)
print([1, 2, 3] + [4, 5])  # concatena
````

### 2.4 Tuplas — sequência imutável
Úteis para valores que não devem mudar e para retornar vários valores de uma função.

````python
coordenada = (-25.43, -49.27)
lat, lon = coordenada      # desempacotamento
print(lat, lon)
````

### 2.5 Dicionários — pares chave → valor
O equivalente a uma lista nomeada do R. Serão a base para criar DataFrames.

````python
aluno = {"nome": "Ana", "idade": 34, "notas": [7.5, 8.0]}
print(aluno["nome"])
print("-------")

aluno["curso"] = "Especialização"     # adiciona chave
print(aluno)
print("-------")

print(aluno.get("email", "não informado"))  # acesso seguro
print("-------")

for chave, valor in aluno.items():
    print(f"{chave}: {valor}")
````

### 2.6 Conjuntos — elementos únicos

````python
cursos = ["ADM", "ECO", "ADM", "ENG", "ECO"]
print(set(cursos))           # elementos únicos, como unique() no R

A = {"ADM", "ENG", "DIR"}
B = {"ENG", "ECO"}

print(A | B)    # união       (R: union(A, B))
print(A & B)    # interseção  (R: intersect(A, B))
print(A - B)    # diferença   (R: setdiff(A, B)): está em A, mas não em B
print(B - A)    # a diferença não é simétrica: a ordem importa
````

## 3. Controle de fluxo e funções

### 3.1 Condicionais
Sem chaves e sem parênteses obrigatórios: o bloco é definido pela **indentação**.

````python
nota = 6.8
if nota >= 7:
    situacao = "Aprovado"
elif nota >= 4:
    situacao = "Exame"
else:
    situacao = "Reprovado"
print(situacao)

# Versão em uma linha (como ifelse para um valor)
print("Aprovado" if nota >= 7 else "Não aprovado")
````

### 3.2 Laços

````python
for i, fruta in enumerate(frutas):   # enumerate dá índice + valor
    print(i, fruta)
````

````python
nomes = ["Ana", "Bruno", "Carla"]
notas = [8.5, 6.0, 9.2]
for n, x in zip(nomes, notas):       # percorre em paralelo
    print(f"{n:<6} {x:>4}")
````

### 3.3 Funções

````python
def classificar(nota, corte=7.0):
    """Retorna a situação do aluno a partir da nota."""
    if nota >= corte:
        return "Aprovado"
    return "Reprovado"

print(classificar(8))
print(classificar(6.5, corte=6))     # argumento nomeado
````

````python
def resumo(valores):
    """Retorna mínimo, média e máximo."""
    return min(valores), sum(valores) / len(valores), max(valores)

mn, med, mx = resumo([4, 8, 6, 10])
print(mn, med, mx)
````

````python
# Funções anônimas (lambda) — como function(x) x^2 no R
quadrado = lambda x: x ** 2
print(quadrado(5))

cidades = ["Curitiba", "Lapa", "Ponta Grossa", "Guarapuava"]
print(sorted(cidades))                           # ordem alfabética
print(sorted(cidades, key=lambda s: len(s)))     # ordena pelo tamanho
````

### 3.4 Comprehensions
A forma "pythônica" de transformar e filtrar coleções — substitui muitos `for` e lembra o `sapply`.

````python
precos = [10.0, 25.5, 8.9, 40.0]

com_desconto = [p * 0.9 for p in precos]
caros = [p for p in precos if p > 20]


print(com_desconto)
print(caros)
````

````python
precos_dict = {f"item_{i}": p for i, p in enumerate(precos)}
print(precos_dict)
````

## 4. NumPy e vetorização

Listas não fazem conta elemento a elemento. O **NumPy** traz o comportamento vetorizado que você conhece do R — e é a base do Pandas.

````python
import numpy as np

v = np.array([1, 2, 3, 4, 5])
print(v * 2)          # agora sim: 2 4 6 8 10
print(v + v)
print(v ** 2)
print(v.mean(), v.std(), v.sum())
````

````python
# Criação de arrays
print(np.arange(0, 10, 2))       # como seq(0, 8, by = 2)
print(np.linspace(0, 1, 5))      # como seq(0, 1, length.out = 5)
print(np.zeros(3), np.ones(3))   # padrão: dtype=float
````

````python
z = np.concatenate([np.zeros(3), np.ones(3)])
print(z)

z = np.concatenate([np.zeros(3, dtype=int), np.ones(3, dtype=int)])
print(z)
````

````python
# Filtro booleano — idêntico em espírito a x[x > 3] no R
v = np.array([3, 8, 1, 9, 4, 7])
print(v > 4)
print(v[v > 4])
print(v[(v > 2) & (v < 8)])      # use & e |, com parênteses
````

````python
# np.where = ifelse()
print(np.where(v >= 5, "alto", "baixo"))
````

````python
# Matrizes (arrays 2D) e eixos
M = np.array([[1, 2, 3],
              [4, 5, 6]])
print(M.shape)
print(M.sum(axis=0))   # soma por coluna  (colSums)
print(M.sum(axis=1))   # soma por linha   (rowSums)
print(M[1, 2])         # linha 1, coluna 2 (lembre: base 0)
````

````python
# Números aleatórios reprodutíveis (R: set.seed)
rng = np.random.default_rng(42)
amostra = rng.normal(loc=70, scale=10, size=1000)
print(amostra[:5].round(2))
print(f"média={amostra.mean():.2f}  dp={amostra.std():.2f}")
````

````python
# Por que vetorizar? Comparando velocidade
import time
x = rng.random(1_000_000)

t0 = time.perf_counter()
s_loop = 0.0
for xi in x:
    s_loop += xi ** 2
t_loop = time.perf_counter() - t0

t0 = time.perf_counter()
s_vec = (x ** 2).sum()
t_vec = time.perf_counter() - t0

print(f"laço: {t_loop:.3f}s | vetorizado: {t_vec:.4f}s | ~{t_loop / t_vec:.0f}x mais rápido")
````

> **Regra prática:** em análise de dados, evite `for` sobre elementos. Prefira operações vetorizadas (NumPy/Pandas). O mesmo vale para o R.

````python
# Exercício 3
def imc(peso, altura):
    valor = round(peso / altura ** 2, 1)
    if valor < 18.5:
        classe = "abaixo"
    elif valor < 25:
        classe = "normal"
    elif valor < 30:
        classe = "sobrepeso"
    else:
        classe = "obesidade"
    return valor, classe

print(imc(70, 1.75))
print(imc(95, 1.70))
````
