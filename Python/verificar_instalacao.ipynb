# **Verificação da instalação**

**Disciplina:** Python e Pandas para Análise de Dados

Execute a célula abaixo (`Shift + Enter`) no JupyterLab do seu computador, **com o ambiente `curso-python` ativo**. Se tudo estiver certo, aparecerão as versões de todos os pacotes e um gráfico com o título "Instalação OK!".

O passo a passo completo está no guia [`INSTALACAO.md`](INSTALACAO.md).

```python
import sys
import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
import seaborn as sns

print(f"Python     {sys.version.split()[0]}")
print(f"NumPy      {np.__version__}")
print(f"Pandas     {pd.__version__}")
print(f"Matplotlib {matplotlib.__version__}")
print(f"Seaborn    {sns.__version__}")
print("Ambiente:", sys.prefix)

x = np.linspace(0, 2 * np.pi, 100)
fig, ax = plt.subplots(figsize=(6, 3))
ax.plot(x, np.sin(x))
ax.set_title("Instalação OK!")
plt.show()
```
