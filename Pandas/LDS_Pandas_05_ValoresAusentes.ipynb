# **Leitura de Dados**

A leitura de dados é o primeiro passo em qualquer fluxo de trabalho de análise de dados. O Pandas facilita a importação de dados de várias fontes e formatos para um DataFrame, que é a estrutura de dados principal utilizada ao longo da análise.

**Arquivos CSV**

O formato CSV (Comma-Separated Values) é um dos formatos de dados mais comuns, utilizado amplamente para armazenar dados tabulares.

```python
import pandas as pd

nba_file = "https://raw.githubusercontent.com/EduPekUfpr/PythonProject/refs/heads/main/Dados/nba.csv" #caminho do arquivo

df_nba = pd.read_csv(nba_file)
print(df_nba.head())  # Exibe as primeiras 5 linhas do DataFrame
```

**Primeira inspeção dos dados**

Antes de caçar valores ausentes, é uma boa prática olhar a estrutura geral do DataFrame: quantidade de linhas/colunas, tipos de dados e um resumo estatístico. Isso já dá pistas de onde podem estar os problemas (colunas com menos valores não-nulos que o total de linhas, por exemplo).

```python
df_nba.info()  # tipos de dados e contagem de valores não-nulos por coluna
df_nba.describe()  # resumo estatístico das colunas numéricas
```

<a id="treat"></a>
# **Tratamento de Valores Ausentes e Duplicados**

Durante a análise de dados, é comum encontrar valores ausentes e dados duplicados. O tratamento adequado desses problemas é essencial para garantir a qualidade dos resultados da análise. O Pandas oferece diversas funções e métodos para identificar, lidar e substituir esses valores de maneira eficiente.

## Identificação de Valores Ausentes

Valores ausentes podem aparecer por vários motivos, como erros de coleta de dados ou ausência real de informação. No Pandas, valores ausentes são representados como `NaN` (Not a Number)

```python
df_nba.isnull() #encontra do DF os valores ausentes. Pode-se usar o isna() também
```

```python
df_nba.isnull().sum() #Somando os booleanos podemos saber a quantidade de ausentes
```

```python
# Contagem absoluta é útil, mas o percentual ajuda a decidir a estratégia:
# poucos % ausentes -> geralmente dá para dropar; muitos % -> dropar destrói a amostra, melhor imputar
(df_nba.isnull().sum() / len(df_nba) * 100).round(2)
```

## Remoção de Valores Ausentes

```python
df_nba_dropRows = df_nba.dropna() #remove as linhas em que alguma coluna tem valor ausente
df_nba_dropRows.isnull().sum()
```

```python
df_nba_dropRows.shape
```

```python
df_nba_dropCols = df_nba.dropna(axis=1) #remove as colunas em que alguma linha tem valor ausente
df_nba_dropCols.isnull().sum()
```

```python
df_nba_dropCols.shape
```

## Substituição de Valores Ausentes

⚠️ **Cuidado:** preencher *todo* o DataFrame com um único valor (como `0`) raramente é uma boa ideia. Colunas numéricas como `Salary` ganhariam um valor sem sentido (salário zero), e colunas de texto como `College` ganhariam um número onde deveria haver uma string. O exemplo abaixo serve só para mostrar **como o método funciona** — na prática, a escolha do valor de preenchimento deve ser feita **coluna por coluna**, como faremos logo em seguida.

```python
#substitui valores nulos por algum valor estabelecido.
#Neste caso ZERO
df_nba.fillna(0)
```

```python
df_nba["Salary"].fillna(df_nba["Salary"].mean())
```

```python
df_nba["College"].fillna("UFPR")
```

```python
df_nba.isnull().sum()
```

**Por que dropar `Salary` mas imputar `College`?**

A decisão não é arbitrária — depende do papel de cada coluna na análise:

- `Salary` é uma variável **numérica central** para qualquer análise financeira do dataset. Inventar um valor para ela (média, zero, etc.) distorceria estatísticas e modelos que a usem como alvo ou feature. Como a proporção de ausentes é pequena (veja o percentual calculado acima), **dropar as linhas** é a opção mais segura.
- `College` é uma variável **categórica** usada mais para agrupamento/contexto do que para cálculo numérico. Preencher com um rótulo distinto (`'UFPR'`) preserva a linha (e o salário, altura, etc. do jogador) sem fingir que sabemos qual foi a faculdade — só marcamos explicitamente que essa informação não estava disponível.

Ou seja: **a estratégia depende de quão crítica a coluna é para a análise e de quanto dado se perderia ao dropar.**

```python
# melhor opção neste caso,
df_nba = df_nba.dropna(subset=['Salary']) # 'dropar' as linhas com valores ausentes na coluna salário
df_nba["College"] = df_nba["College"].fillna("UFPR") # Substituir por um valor diferente dos demais na coluna 'college'
df_nba.isnull().sum()
df_nba.shape
```

```python
df_nba.isnull().sum()
```

<a id="dup"></a>
## Identificação e Remoção de Duplicados

Além de valores ausentes, é comum encontrar **linhas duplicadas** em um dataset — registros repetidos por erro de coleta, junções (`merge`) mal feitas, ou reenvio do mesmo dado. O Pandas oferece `duplicated()` para identificar e `drop_duplicates()` para remover.

```python
df_nba.duplicated().sum()  # quantidade de linhas totalmente duplicadas
```

Se houver duplicados, vale inspecionar quais linhas são antes de decidir removê-las — às vezes uma aparente duplicata é, na verdade, um dado legítimo (por exemplo, dois jogadores diferentes com o mesmo nome). Por padrão, `duplicated()` marca como `True` a partir da **segunda** ocorrência em diante.

```python
df_nba[df_nba.duplicated(keep=False)]  # mostra TODAS as ocorrências envolvidas em duplicação (não só a repetida)
```

Também é possível considerar duplicado com base em **algumas colunas específicas**, não a linha inteira — útil quando, por exemplo, duas linhas têm o mesmo jogador (`Name`) mas com pequenas diferenças em outras colunas por erro de digitação.

```python
# Exemplo: considerar duplicado apenas pelo nome do jogador
df_nba.duplicated(subset=['Name']).sum()
```

```python
# Remove duplicados, mantendo a primeira ocorrência (padrão: keep='first')
df_nba = df_nba.drop_duplicates()
df_nba.shape
```
